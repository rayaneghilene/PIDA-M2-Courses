## Let's use our new coding skills to run an opensource model


The `statement` column contains text that we can read. This code runs a named entity recognition model on each statement and adds its predictions to a new `entities` column. In Lecture 4, we will use that column to ask questions about the people, organizations, and places mentioned in the dataset.


Each `entities` cell contains a list of predictions, including the entity text (`word`), type (`entity_group`), and score. An empty list means the model found no entities.

## Named-Entity Recognition (NER)

```python
from transformers import pipeline

ner = pipeline(
    "token-classification",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple",
)

results = []

for statement in df["statement"]:
    entities = ner(statement) if isinstance(statement, str) and statement.strip() else []
    results.append(entities)

df["entities"] = results

df[["statement", "entities"]].head()
```
